## 1.2 – XÂY DỰNG CÁC HÀM XỬ LÝ
#### Mục tiêu
Tạo các hàm xử lý dữ liệu theo yêu cầu của ban giám đốc.
#### Giải thích
Các hàm dưới đây được thiết kế theo hướng tái sử dụng để phục vụ:
- Phần OOP
- Phần NumPy/Pandas
- Phần Visualization
- Phần Dashboard

```python
# a, hàm lọc đơn hàng
def loc_don_hang(ds, **tieu_chi):

    ket_qua = []

    for order in ds:
        valid = True

        for key, value in tieu_chi.items():
            if order.get(key) != value:
                valid = False
                break

        if valid:
            ket_qua.append(order)

    return ket_qua
# ví dụ minh họa 
hn_completed = loc_don_hang(
    orders,
    khu_vuc="Hà Nội",
    trang_thai="Hoàn thành"
)

print("Số đơn:", len(hn_completed))
```

```python
# b, hàm thống kê theo nhóm
def thong_ke_theo_nhom(ds, truong_nhom):

    result = {}

    groups = {order[truong_nhom] for order in ds}

    for group in groups:

        data = [
            order
            for order in ds
            if order[truong_nhom] == group
        ]

        so_don = len(data)

        tong_gia_tri = sum(
            order["gia_tri"]
            for order in data
        )

        gia_tri_tb = tong_gia_tri / so_don

        completed = len(
            [
                order
                for order in data
                if order["trang_thai"] == "Hoàn thành"
            ]
        )

        result[group] = {
            "so_don": so_don,
            "tong_gia_tri": tong_gia_tri,
            "gia_tri_tb": round(gia_tri_tb, 0),
            "ty_le_hoan_thanh": round(
                completed / so_don * 100,
                2
            )
        }

    return result
# ví dụ minh họa
region_report = thong_ke_theo_nhom(
    orders,
    "khu_vuc"
)

for region, stats in region_report.items():
    print(region, stats)
```

```python
# c, hàm phân khúc đơn hàng
def phan_khuc_don_hang(ds):

    for order in ds:

        value = order["gia_tri"]

        if value < 5_000_000:
            segment = "Thấp"

        elif value < 15_000_000:
            segment = "Trung bình"

        else:
            segment = "Cao"

        order["phan_khuc"] = segment
# ví dụ minh họa
phan_khuc_don_hang(orders)

for order in orders[:5]:
    print(
        order["ma_don"],
        order["phan_khuc"]
    )
```

```python
# d, hàm executive summary
def bao_cao_tom_tat(ds):

    total_orders = len(ds)

    total_revenue = sum(
        order["gia_tri"]
        for order in ds
    )

    avg_value = total_revenue / total_orders

    max_order = max(
        ds,
        key=lambda x: x["gia_tri"]
    )

    min_order = min(
        ds,
        key=lambda x: x["gia_tri"]
    )

    completed_orders = len(
        [
            order
            for order in ds
            if order["trang_thai"] == "Hoàn thành"
        ]
    )

    completion_rate = (
        completed_orders /
        total_orders * 100
    )

    top_region = max(
        thong_ke_theo_nhom(ds, "khu_vuc").items(),
        key=lambda x: x[1]["tong_gia_tri"]
    )[0]

    top_channel = max(
        thong_ke_theo_nhom(ds, "kenh_ban").items(),
        key=lambda x: x[1]["tong_gia_tri"]
    )[0]

    top_category = max(
        thong_ke_theo_nhom(ds, "danh_muc").items(),
        key=lambda x: x[1]["tong_gia_tri"]
    )[0]

    print("PHENIKAA COMMERCE - EXECUTIVE SUMMARY")

    print(f"Tổng doanh thu      : {total_revenue:,.0f} VND")
    print(f"Tổng số đơn         : {total_orders}")
    print(f"Giá trị TB          : {avg_value:,.0f} VND")
    print(f"Đơn lớn nhất        : {max_order['ma_don']}")
    print(f"Đơn nhỏ nhất        : {min_order['ma_don']}")
    print(f"Top khu vực         : {top_region}")
    print(f"Top kênh bán        : {top_channel}")
    print(f"Top danh mục        : {top_category}")
    print(f"Số đơn hoàn thành   : {completed_orders}")
    print(f"Tỷ lệ hoàn thành    : {completion_rate:.2f}%")
    
bao_cao_tom_tat(orders)
```

Việc thành thạo các kỹ thuật này giúp giảm số lượng vòng lặp thủ công, làm cho chương trình dễ đọc hơn và chuẩn bị tốt cho việc chuyển sang Pandas ở các phần tiếp theo.

### a, Filter() – Lọc đơn hàng giá trị lớn hơn 10 triệu
#### Mục đích
Ban giám đốc muốn nhanh chóng xác định các đơn hàng có giá trị cao để theo dõi nhóm khách hàng doanh nghiệp quan trọng.

Filter() cho phép lọc dữ liệu theo điều kiện mà không cần viết vòng lặp dài dòng.

```python
# Lọc đơn hàng có giá trị trên 10 triệu VND

high_value_orders = list(
    filter(
        lambda order: order["gia_tri"] > 10_000_000,
        orders
    )
)

print("DANH SÁCH ĐƠN HÀNG GIÁ TRỊ CAO (>10 TRIỆU)")

for order in high_value_orders:
    print(
        order["ma_don"],
        f"{order['gia_tri']:,.0f} VND"
    )

print("\nTổng số đơn:", len(high_value_orders))
```

#### Nhận xét 

- Có 10 đơn hàng có giá trị trên 10 triệu đồng.
- Đây là nhóm đơn hàng tạo ra phần lớn doanh thu công ty.
- Kỹ thuật filter() giúp lọc dữ liệu nhanh và dễ đọc hơn so với sử dụng nhiều câu lệnh if.

### b, Map() – Trích xuất danh sách doanh thu
#### Mục đích
Trong các bước phân tích tiếp theo, nhóm dự án cần danh sách giá trị đơn hàng để tính toán doanh thu, trung bình, độ lệch hoặc chuyển sang NumPy.

map() cho phép biến đổi dữ liệu từ cấu trúc phức tạp sang dữ liệu đơn giản hơn.

```python
# Lấy danh sách doanh thu từng đơn hàng

revenues = list(
    map(
        lambda order: order["gia_tri"],
        orders
    )
)

print("DANH SÁCH DOANH THU")

print(revenues)
```

#### Nhận xét

- map() được sử dụng để lấy riêng trường gia_tri.
- Kết quả là một danh sách số nguyên dễ tính toán.
- Danh sách này sẽ được tái sử dụng ở phần NumPy và Pandas sau này.

### c, Sorted() + Lambda – Sắp xếp doanh thu giảm dần
#### Mục đích
- Nhà quản trị thường muốn biết đơn hàng nào mang lại doanh thu cao nhất.
- sorted() kết hợp với lambda giúp sắp xếp linh hoạt theo bất kỳ thuộc tính nào trong Dictionary.

```python
# Sắp xếp đơn hàng theo doanh thu giảm dần

sorted_orders = sorted(
    orders,
    key=lambda order: order["gia_tri"],
    reverse=True
)

print("TOP 5 ĐƠN HÀNG GIÁ TRỊ CAO NHẤT")

for order in sorted_orders[:5]:

    print(
        order["ma_don"],
        "-",
        f"{order['gia_tri']:,.0f} VND"
    )
```

- lambda giúp xác định tiêu chí sắp xếp ngắn gọn và linh hoạt.

#### Mục đích

# PHẦN 2 – LẬP TRÌNH HƯỚNG ĐỐI TƯỢNG (OOP)
## 2.1 Thiết kế hệ thống lớp cho Phenikaa Commerce
#### Mục tiêu
Xây dựng hệ thống quản lý khách hàng theo hướng đối tượng nhằm hỗ trợ phân tích hành vi mua hàng và triển khai các chính sách chăm sóc khách hàng.
#### Giải thích
Trong hệ thống thương mại điện tử, khách hàng thường được chia thành nhiều nhóm khác nhau. OOP giúp mô hình hóa dữ liệu thực tế thành các đối tượng có thuộc tính và hành vi riêng.
#### Thiết kế gồm:
- KhachHang (lớp cha)
- KhachHangVIP (lớp con kế thừa)
- SanPham (lớp độc lập)
### 2.1.1 Xây dựng lớp KhachHang

```python
class KhachHang:
    
    def __init__(
        self,
        ma_kh,
        ho_ten,
        email,
        dia_chi,
        ngay_dang_ky
    ):
        self.ma_kh = ma_kh
        self.ho_ten = ho_ten
        self.email = email
        self.dia_chi = dia_chi
        self.ngay_dang_ky = ngay_dang_ky

    @property
    def email(self):
        return self._email

    @email.setter
    def email(self, value):

        if "@" not in value:
            raise ValueError(
                "Email không hợp lệ!"
            )

        self._email = value

    def lich_su_mua_hang(
        self,
        ds_don_hang
    ):
        return [
            order
            for order in ds_don_hang
            if order["khu_vuc"] == self.dia_chi
        ]

    def __str__(self):
        return (
            f"{self.ma_kh} | "
            f"{self.ho_ten} | "
            f"{self.email}"
        )

    def __lt__(self, other):
        return self.ho_ten < other.ho_ten
```

#### Nhận xét

- Sử dụng @property và setter để kiểm tra email.
- Dùng __lt__() để hỗ trợ sắp xếp bằng sorted().
- lich_su_mua_hang() liên kết trực tiếp với dữ liệu đơn hàng của Phần 1.
### 2.1.2 Xây dựng lớp KhachHangVIP


```python
class KhachHangVIP(KhachHang):

    def __init__(
        self,
        ma_kh,
        ho_ten,
        email,
        dia_chi,
        ngay_dang_ky,
        hang_vip,
        diem_tich_luy
    ):

        super().__init__(
            ma_kh,
            ho_ten,
            email,
            dia_chi,
            ngay_dang_ky
        )

        self.hang_vip = hang_vip
        self.diem_tich_luy = diem_tich_luy

    def tinh_uu_dai(self):

        if self.hang_vip == "Silver":
            return 0.05

        elif self.hang_vip == "Gold":
            return 0.10

        elif self.hang_vip == "Platinum":
            return 0.15

        return 0

    def __str__(self):

        return (
            f"{self.ma_kh} | "
            f"{self.ho_ten} | "
            f"{self.hang_vip} | "
            f"{self.diem_tich_luy} điểm"
        )
```

#### Nhận xét

- Kế thừa từ lớp KhachHang.
- Sử dụng super().__init__().
- Ghi đè phương thức __str__().
- Bổ sung phương thức tính ưu đãi theo hạng VIP
### 2.1.3 Xây dựng lớp SanPham


```python
class SanPham:

    VAT_RATE = 0.10

    def __init__(
        self,
        ma_sp,
        ten,
        danh_muc,
        gia,
        ton_kho,
        muc_canh_bao
    ):

        self.ma_sp = ma_sp
        self.ten = ten
        self.danh_muc = danh_muc
        self.gia = gia
        self.ton_kho = ton_kho
        self.muc_canh_bao = muc_canh_bao

    def can_nhap_hang(self):

        return self.ton_kho <= self.muc_canh_bao

    def tinh_gia_sau_thue(self):

        return self.gia * (
            1 + self.VAT_RATE
        )

    def __repr__(self):

        return (
            f"SanPham("
            f"{self.ma_sp}, "
            f"{self.ten}, "
            f"{self.gia:,.0f} VND)"
        )

    def __eq__(
        self,
        other
    ):
        return (
            self.ma_sp == other.ma_sp
        )
```

#### Nhận xét

- Có __repr__() để biểu diễn đối tượng.
- Có __eq__() để so sánh sản phẩm.
- Có chức năng kiểm tra nhập hàng và tính giá sau thuế.

# PHẦN 3 – XỬ LÝ DỮ LIỆU VỚI NUMPY & PANDAS
## 3.1 Tạo và kiểm tra dữ liệu
#### Mục tiêu
Xây dựng tập dữ liệu kinh doanh quy mô lớn phục vụ phân tích dữ liệu.
#### Giải thích
Trong thực tế, dữ liệu thường có hàng trăm hoặc hàng nghìn bản ghi. Vì vậy phần này tạo DataFrame mô phỏng 300 đơn hàng để thực hiện EDA, thống kê và trực quan hóa ở các phần sau.

```python
%pip install pandas
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
```

```python
np.random.seed(2026)

N = 300

df_orders = pd.DataFrame({
    "ma_don":
        [f"DH{i:04d}" for i in range(1, N + 1)],

    "ngay_dat":
        pd.date_range(
            start="2025-01-01",
            periods=N,
            freq="D"
        ),

    "khu_vuc":
        np.random.choice(
            [
                "Hà Nội",
                "TP.HCM",
                "Đà Nẵng",
                "Hải Phòng",
                "Cần Thơ"
            ],
            N
        ),

    "kenh_ban":
        np.random.choice(
            [
                "Website",
                "Sales",
                "Đối tác",
                "Shopee",
                "Lazada"
            ],
            N
        ),

    "danh_muc":
        np.random.choice(
            [
                "Laptop",
                "Server",
                "Máy in",
                "Router",
                "Thiết bị mạng"
            ],
            N
        ),

    "gia_tri":
        np.random.randint(
            2_000_000,
            50_000_001,
            N
        ),

    "trang_thai":
        np.random.choice(
            [
                "Hoàn thành",
                "Đang xử lý",
                "Đã hủy",
                "Hoàn tiền"
            ],
            N,
            p=[0.65, 0.15, 0.15, 0.05]
        )
})
```

#### Tạo lỗi dữ liệu để thực hành làm sạch và Kiểm tra dữ liệu

```python
# Tạo NaN

df_orders.loc[10, "gia_tri"] = np.nan
df_orders.loc[50, "kenh_ban"] = np.nan

# Tạo dòng trùng

df_orders = pd.concat(
    [
        df_orders,
        df_orders.iloc[[0]]
    ],
    ignore_index=True
)

# Tạo outlier

df_orders.loc[100, "gia_tri"] = 250_000_000
print("Kích thước dữ liệu:")
print(df_orders.shape)

print("\nSố giá trị thiếu:")
print(df_orders.isnull().sum())

print("\nSố dòng trùng:")
print(df_orders.duplicated().sum())
```

#### Làm sạch dữ liệu

```python
print("BEFORE CLEANING")
print(df_orders.shape)

# Điền giá trị thiếu

df_orders["gia_tri"] = (
    df_orders["gia_tri"]
    .fillna(
        df_orders["gia_tri"].median()
    )
)

df_orders["kenh_ban"] = (
    df_orders["kenh_ban"]
    .fillna("Website")
)

# Xóa trùng

df_orders = (
    df_orders
    .drop_duplicates()
)

print("\nAFTER CLEANING")
print(df_orders.shape)
```

#### Nhận xét

- Dữ liệu đã được xử lý NaN.
- Dòng trùng lặp đã được loại bỏ.
- Tập dữ liệu sẵn sàng cho EDA.
## 3.2 PHÂN TÍCH DỮ LIỆU ĐA CHIỀU
### 3.2.1. Thống kê mô tả
#### Mục tiêu
Khảo sát nhanh phân phối của biến doanh thu đơn hàng.

```python
print(
    df_orders[
        ["gia_tri"]
    ].describe()
)
```

#### Nhận xét

- Mean thể hiện giá trị đơn trung bình.
- Median (50%) giúp đánh giá phân phối.
- Max giúp phát hiện outlier.
- Std thể hiện mức độ biến động.
### 3.2.2. Phân tích theo kênh bán

```python
channel_report = (
    df_orders
    .groupby("kenh_ban")
    .agg(
        so_don=("ma_don", "count"),
        tong_dt=("gia_tri", "sum"),
        dt_tb=("gia_tri", "mean"),
        ty_le_done=(
            "trang_thai",
            lambda x:
                (
                    x.eq("Hoàn thành")
                    .mean()
                    * 100
                )
        )
    )
    .round(2)
    .sort_values(
        "tong_dt",
        ascending=False
    )
)

channel_report
```

#### Nhận xét

- Xác định kênh doanh thu cao nhất.
- So sánh hiệu quả giữa các kênh bán hàng.

### 3.2.3. Phân tích xu hướng theo thời gian

```python
df_done = df_orders[
    df_orders["trang_thai"]
    == "Hoàn thành"
].copy()

df_done["thang"] = (
    df_done["ngay_dat"]
    .dt.month
)

monthly_sales = (
    df_done
    .groupby("thang")
    ["gia_tri"]
    .sum()
    / 1e6
)

rolling_3m = (
    monthly_sales
    .rolling(3)
    .mean()
)

print(monthly_sales)

print("\nRolling Average:")
print(rolling_3m)

print(
    "\nTháng cao nhất:",
    monthly_sales.idxmax()
)

print(
    "Tháng thấp nhất:",
    monthly_sales.idxmin()
)
```

#### Nhận xét

- Rolling Average giúp làm mượt dữ liệu.
- Xác định mùa cao điểm và thấp điểm.

### 3.2.4. Phân tích địa lý

```python
pivot_region = (
    pd.pivot_table(
        df_orders,
        values="gia_tri",
        index="khu_vuc",
        columns="danh_muc",
        aggfunc="sum",
        fill_value=0
    )
)

pivot_region
```

#### Tỷ lệ đóng góp doanh thu

```python
region_percent = (
    df_orders
    .groupby("khu_vuc")
    ["gia_tri"]
    .sum()
)

region_percent = (
    region_percent
    / region_percent.sum()
    * 100
)

print(
    region_percent.round(2)
)
```

#### Nhận xét

- Pivot Table giúp quan sát doanh thu theo khu vực × danh mục.
- Tỷ lệ đóng góp hỗ trợ quyết định phân bổ nguồn lực.
### 3.2.5. Phân tích tương quan Pearson

```python
df_orders["thang"] = (
    df_orders["ngay_dat"]
    .dt.month
)

df_orders["gia_tri_nghin"] = (
    df_orders["gia_tri"]
    / 1000
)

df_orders["phi_van_chuyen"] = (
    df_orders["gia_tri"]
    * 0.02
)

corr_matrix = (
    df_orders[
        [
            "gia_tri",
            "gia_tri_nghin",
            "phi_van_chuyen",
            "thang"
        ]
    ]
    .corr()
)

print(corr_matrix)
```

#### Nhận xét

- Pearson gần +1 → tương quan thuận mạnh.
- Pearson gần -1 → tương quan nghịch mạnh.
- Pearson gần 0 → không có tương quan.

### 3.3 PHÂN TÍCH NUMPY TRỰC TIẾP
#### Mục tiêu
Thực hành vectorization với NumPy mà không dùng vòng lặp.

### 3.3.1. Tính doanh thu sau chiết khấu

```python
sales = (
    df_orders["gia_tri"]
    .to_numpy()
)

discounted_sales = (
    sales * 0.95
)

print(
    discounted_sales[:10]
)
```

#### Nhận xét

- Toàn bộ phép tính áp dụng trên mảng.
- Không sử dụng vòng lặp.
- Chính là Vectorization. 
### 3.3.2. Phân loại đơn hàng bằng np.where()

```python
segments = np.where(
    sales < 10_000_000,
    "Thấp",
    np.where(
        sales < 30_000_000,
        "Trung bình",
        "Cao"
    )
)

print(
    segments[:20]
)
```

#### Nhận xét

- np.where() hoạt động tương tự if-else.
- Tốc độ xử lý nhanh trên dữ liệu lớn.
### 3.3.3. Xác định phân vị và outlier bằng IQR


```python
q1 = np.percentile(
    sales,
    25
)

q2 = np.percentile(
    sales,
    50
)

q3 = np.percentile(
    sales,
    75
)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = sales[
    (sales < lower)
    |
    (sales > upper)
]

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)

print(
    "\nSố outlier:",
    len(outliers)
)
```

Dashboard gồm:

- Xu hướng doanh thu theo tháng
- Doanh thu theo khu vực
- Cơ cấu trạng thái đơn hàng
- Tương quan giá trị đơn hàng và phí vận chuyển

```python
import matplotlib.pyplot as plt
from matplotlib import gridspec
import numpy as np
```

#### Chuẩn bị dữ liệu Dashboard

```python
# Chỉ lấy đơn hoàn thành

df_done = df_orders[
    df_orders["trang_thai"]
    == "Hoàn thành"
].copy()

# Doanh thu theo tháng

monthly_sales = (
    df_done
    .groupby(
        df_done["ngay_dat"].dt.month
    )["gia_tri"]
    .sum()
    / 1e6
)

rolling_3 = (
    monthly_sales
    .rolling(3)
    .mean()
)

# Doanh thu khu vực

region_sales = (
    df_done
    .groupby("khu_vuc")
    ["gia_tri"]
    .sum()
    / 1e6
)

# Trạng thái đơn hàng

status_count = (
    df_orders["trang_thai"]
    .value_counts()
)

# Scatter Data

x = df_orders["gia_tri"] / 1e6

y = df_orders["phi_van_chuyen"] / 1000
```

#### Dashboard Matplotlib

```python
fig = plt.figure(
    figsize=(16, 10)
)

gs = gridspec.GridSpec(
    2,
    2,
    hspace=0.35,
    wspace=0.30
)

# ====================
# AX1 LINE CHART
# ====================

ax1 = fig.add_subplot(gs[0, 0])

ax1.plot(
    monthly_sales.index,
    monthly_sales.values,
    "bo-",
    lw=2,
    label="Doanh thu"
)

ax1.fill_between(
    monthly_sales.index,
    monthly_sales.values,
    alpha=0.20
)

ax1.plot(
    rolling_3.index,
    rolling_3.values,
    "r--",
    lw=2,
    label="Rolling Avg 3T"
)

peak_month = monthly_sales.idxmax()
peak_value = monthly_sales.max()

ax1.annotate(
    f"Đỉnh: {peak_value:.0f}M",
    xy=(peak_month, peak_value),
    xytext=(
        peak_month - 1,
        peak_value + 150
    ),
    arrowprops={
        "arrowstyle": "->"
    }
)

ax1.set_title(
    "A. Xu hướng doanh thu theo tháng"
)

ax1.set_xlabel("Tháng")
ax1.set_ylabel("Triệu VND")

ax1.grid(True)

ax1.legend()

# ====================
# AX2 BAR CHART
# ====================

ax2 = fig.add_subplot(gs[0, 1])

bars = ax2.bar(
    region_sales.index,
    region_sales.values,
    color="steelblue"
)

ax2.set_title(
    "B. Doanh thu theo khu vực"
)

ax2.set_xlabel("Khu vực")

ax2.set_ylabel(
    "Triệu VND"
)

for bar in bars:

    h = bar.get_height()

    ax2.text(
        bar.get_x()
        + bar.get_width()/2,
        h,
        f"{h:.0f}",
        ha="center"
    )

# ====================
# AX3 PIE CHART
# ====================

ax3 = fig.add_subplot(gs[1, 0])

ax3.pie(
    status_count.values,
    labels=status_count.index,
    autopct="%1.1f%%",
    startangle=90
)

ax3.set_title(
    "C. Cơ cấu trạng thái đơn hàng"
)

# ====================
# AX4 SCATTER
# ====================

ax4 = fig.add_subplot(gs[1, 1])

ax4.scatter(
    x,
    y,
    alpha=0.6,
    color="orange"
)

z = np.polyfit(
    x,
    y,
    1
)

p = np.poly1d(z)

ax4.plot(
    x,
    p(x),
    "r--",
    lw=2,
    label="Trend Line"
)

ax4.set_title(
    "D. Giá trị đơn hàng và phí vận chuyển"
)

ax4.set_xlabel(
    "Giá trị đơn (Triệu VND)"
)

ax4.set_ylabel(
    "Phí vận chuyển (Nghìn VND)"
)

ax4.legend()

ax4.grid(True)

fig.suptitle(
    "PHENIKAA COMMERCE BUSINESS DASHBOARD 2026",
    fontsize=16,
    fontweight="bold"
)


plt.savefig(
    "dashboard_matplotlib.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

#### Nhận xét

- Line Chart cho thấy xu hướng doanh thu theo tháng.
- Rolling Average giúp làm mượt biến động.
- Bar Chart hỗ trợ so sánh khu vực.
- Pie Chart cho thấy cơ cấu trạng thái đơn hàng.

# PHẦN 5 – BÁO CÁO TỔNG KẾT VÀ ĐỀ XUẤT
## 5.1 TÓM TẮT DỰ ÁN
#### Mục tiêu
Tổng hợp các chỉ số kinh doanh quan trọng giúp Ban Giám đốc nhanh chóng đánh giá hiệu quả hoạt động của Phenikaa Commerce.
#### Giải thích
Executive Summary là phần tóm tắt ngắn gọn các KPI quan trọng nhất của doanh nghiệp trước khi đưa ra quyết định quản trị.

```python
print("PHENIKAA COMMERCE - EXECUTIVE SUMMARY")

# KPI tổng quan

total_revenue = (
    df_orders["gia_tri"]
    .sum()
)

completed_orders = (
    df_orders["trang_thai"]
    .eq("Hoàn thành")
    .sum()
)

total_orders = len(df_orders)

aov = (
    total_revenue
    / total_orders
)

cancel_rate = (
    (
        df_orders["trang_thai"]
        == "Đã hủy"
    ).mean()
    * 100
)

print(f"Tổng doanh thu            : {total_revenue:,.0f} VND")
print(f"Tổng số đơn hàng          : {total_orders}")
print(f"Số đơn hoàn thành         : {completed_orders}")
print(f"AOV trung bình            : {aov:,.0f} VND")
print(f"Tỷ lệ hủy đơn             : {cancel_rate:.2f}%")
```

#### Điểm nổi bật của dữ liệu

```python
# Top khu vực

top_regions = (
    df_orders
    .groupby("khu_vuc")["gia_tri"]
    .sum()
    .sort_values(ascending=False)
)

# Top kênh

top_channels = (
    df_orders
    .groupby("kenh_ban")["gia_tri"]
    .sum()
    .sort_values(ascending=False)
)

# Tháng doanh thu cao nhất

best_month = (
    monthly_sales.idxmax()
)

best_month_value = (
    monthly_sales.max()
)

print("ĐIỂM NỔI BẬT")

print("\nTOP 3 KHU VỰC HIỆU QUẢ NHẤT")

print(top_regions.head(3))

print("\nTOP 3 KÊNH BÁN HIỆU QUẢ NHẤT")

print(top_channels.head(3))

print(
    f"\nTháng bứt phá nhất: "
    f"Tháng {best_month}"
)

print(
    f"Doanh thu: "
    f"{best_month_value:.0f} triệu VND"
)
```
